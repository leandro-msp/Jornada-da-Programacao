```python
import numpy as np 
from sklearn.linear_model import LinearRegression
```

```python
#SIMULAÇAO DE Previsão de vendas e comparativo de média de vendas por funcionário

anúncios = ([[10],[20],[40],[60]])
vendas = ([20,45,62,94])

modelo = LinearRegression()
modelo.fit(anúncios,vendas)

nova_quantidade = [[250]]
previsao_vendas = modelo.predict(nova_quantidade)

print(f"A estimativa de vendas será de {int(previsao_vendas[0])}, caso a equipe de publicidade atinja o total de {nova_quantidade[0][0]} anúncios.")

total_vendas = int(previsao_vendas[0])

func = ['Juliana','Marcos','Cláuida','Leandro','Rafaela'] 

qtd_func = len(func) #contagem da quantiade de elementos dentro da lista func (funcionários)

# print(qtd_func)

media_vendas = total_vendas/qtd_func

print(f"Com esse total de vendas({int(total_vendas)}), a média de vendas por funcionário será: {round(media_vendas)}")


with open("relatorio_vendas.txt","w", encoding="utf-8") as arquivo:
    #abre/cria novo arquivo com o nome definido, com formatação para não quebrar textos com acentos, e o apelida como 'arquivo' para passar para os futuros tratamentos 
    arquivo.write("=== RELATÓRIO DE PREVISÃO DE VENDAS ===\n")
    arquivo.write(f"Total de anúncios planejados: {nova_quantidade[0][0]}\n")
    arquivo.write(f"Estimativa de vendas total: {int(total_vendas)}\n")
    arquivo.write(f"Quantidade de funcionários: {qtd_func}\n")
    arquivo.write(f"Média de vendas por funcionário: {round(media_vendas)}\n")
print("Relatório registrao com sucesso em 'relatório_vendas.txt'!")
```
