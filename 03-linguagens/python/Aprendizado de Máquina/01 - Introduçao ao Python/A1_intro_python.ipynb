```python
#Variáveis
    # a= 5 -> Int 
    # b= 5.0 -> Float 
    # c= "5" -> String 
    # d= true/false -> boolean 
    
x=5
y=10 
# x,y são variáveis do tipo inteiro

resultado = (x>10) and (y<20) # o operador AND/E requer que ambas condições(comparações) sejam verdadeiras.

#será atribuído para a variável "resultado" , a saída da operação comparativa e lógica das variáveis x e y.
resultado

#o resultado é false, pois x(5)  é maior que 10, ou seja false, e y(10) é menor que 20, ou seja, true. Porém o operador AND exige que ambas sejam verdadeiras.
```

```python
x=5
y=25
resultado = (x==5) or (y<20)  # o operador OR/OU necessita que apenas uma condição seja verdadeira, para que o resultado seja True.

#x(5) que é igual a 5, True | y(25), não é menor que 20, False.  Porém uma sendo verdadeira, já passa a ser válida.

resultado
```

```python
x=10
resultado=not(x>20) #operador NOT inverte o valor da operação(condição), ou seja, se uma operação tiver como resultado um valor Booleano True, passará a ser False, e vice-versa.

#x= 10, que não é maior que 20, ou seja, False. Porém com o operador NOT passa a ser True.
# em outra análise lógica seria o mesmo que dizer "10 Não é maior que 20", True
resultado
```

```python
# ESTRUTURA CONDICIONAL -- > O código possue um bloco de instrução do que será realizado caso uma das condições for atendida.

idade=20

if idade >=18: # if = Se    
  print("vc é maior de idade")
  
  # --> se/caso a varíavel idade for maior ou igual a 18, o código realizará a instrução acima (Gerar a saída com o texto definido)
  
else: # else = caso contrário
  print("vc é menor de idade")
  
  # --> se a primeira condição não for atendida(if),a instrução desde bloco(else) será realizada.
```

```python
#ESTRUTURA DE REPETIÇÃO | LAÇO FOR

nomes = ["Ana", "João","Maria","Pedro"]  # Arrays --- > uma variável que pode armazenar mais de um valor e de tipos diferentes.

for x in nomes: 
    # for = uma estrutura que cria uma variável para contar as "voltas", uma condição para definir até quando o ciclo acontecerá, e o passo, que aumenta ou diminui o valor  a cada volta.
    # para cada vez que ele rodar a lista ele irá atribuir para X cada nome dentro da lista
    # for é utilizado quando já se tem um valor fixo(predefinido) de elementos.
    
  print ("Olá" , x) 
```

```python
contador =  5
while contador >0: 
    ''' while = enquanto | também é uma estrutura de repetição, porém com condição que nn tem limitei fixo. Ela define que um bloco de instrução vai ser executado,
    mas somente enquanto a condição que está sendo analisada for verdadeira.'''
    
    print(contador)
    contador -=1
print ("Decolar!")
```
