```python
import numpy as np 
from sklearn.linear_model import LinearRegression
```

```python
#Treinando Modelo de previsão de Temperatura


# grade de valores indexados
horas = np.array([[5],[8],[9],[10],[11],[14],[16],[18],[20],[21],[22]])  
temperatura = np.array([15,20,25,28,30,32,30,27,25,22,19])

#Criando e Treinando o Modelo


modelo = LinearRegression() # aqui define que o modelo será de Regressão Linear
modelo.fit(horas,temperatura) # com modelo de Regressão Linear definido, a função fit treina o modelo com base nos parâmetros passados (base de dados)

nova_hora = [[23]] #novo elemento que ainda não se encontra na grade valores(valor teste)

previsao = modelo.predict(nova_hora) 
# a variável "previsao" receberá a atribuição do resultado da função PREDICT. 
# O predict é função de prever qual é valor do elemento teste passado, com base no treinamento feito a partir da base de dados.

print(f"Temperatura ás 23h será: {previsao}")
```

```python
hora_estudo = np.array([[1],[2],[3],[7],[10]])
nota = np.array ([2,5,8,8,1])

modelo1 = LinearRegression()
modelo1.fit(hora_estudo,nota)

nova_horario = [[9]]
previsao = modelo1.predict(nova_horario)
print(f"Nota caso estude 9 horas: {previsao}")

'''--> o modelo quando passa pelo treinamento, ele se adapta conforme a base de dados de teste que foi passada,
neste cenário, a base de dados apresenta uma queda drástica na nota conforme o aumento de horas, 3 e 7 horas = nota 8 | 10 horas = nota 1
sendo assim o modelo foi treinado com base em: "mais horas, menos desempenho"
'''
```

```python
#Preço de uma Casa considerando apenas o fator do Tamanho:


tamanho = np.array([[40],[60],[90],[120],[140]])
preço = np.array([70000,85000,104000,130000,150000])

modelo = LinearRegression()
modelo.fit(tamanho,preço)

novo_tamanho = [[150]]
previsao = modelo.predict(novo_tamanho)

print(f"Uma casa com 150m² custará: {previsao} ") 
```
