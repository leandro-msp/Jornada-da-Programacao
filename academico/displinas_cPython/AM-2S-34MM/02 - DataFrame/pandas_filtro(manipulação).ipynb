```python
import pandas as pd
```

Utiizando operadores para formar condições e filtrar saída de dados.

```python
dados = {
    'Produto' : ['Celular','TV','Fone de ouvido','Microfone'],
    'Valor' : [ 3000,5000,120,2000],
    'Quantidade' : [50,30,40,5]
}

df = pd.DataFrame(dados)

filtro = df[(df['Valor']>150) &(df['Quantidade'] >10)] # resgatar valores cujo o Valor seja maior que 150 E(operador and) Quantidade seja maior que 10.
    # Está sendo utilizado o operador AND, ou seja ambas condições devem ser verdadeiras, caso só uma for, nenhum valor será mostrado na saída.
    
print("Valores maiores que 150 e quantidade maior que 10:")
filtro
```

```python
# filtrar informação dos produtos cujo a Quantidade seja inferior a 50

dados = {
    'Produto' : ['Caderno','Lápis','Caneta','Estojo'],
    "Quantidade" : [10,27,54,4]
}

df = pd.DataFrame(dados)

filtro = df[df['Quantidade']<50] 
print ("Produtos com quantidade abaixo de 50:")
filtro
```

```python
# filtrar informação dos funcionários cujo o Departamento seja de Vendas

dados = {
    'Nome' : ['João','Maria','Milene','Richard'],
    'Salário' :[2000,1600,800,2500],
    'Departamento': ['Vendas','RH','Financeiro','Vendas']

}

df = pd.DataFrame(dados)
filtro = df[df['Departamento']=='Vendas']  
filtro
```

```python
#filtrar informações dos Alunos cujo a idade seja igual ou maior a 28.

dados = {
    'Nome' : ['João','Maria','Ana','Roberta','Letícia'],
    'Curso' : ['TI','Artes','Direito','Pedagogia','Ed.Física'],
    'Idade' : [25,28,35,42,18]
}

df = pd.DataFrame(dados)
filtro = df[df['Idade']>=28] 
filtro
```

```python
# filtrar os medicamentos que tenham menos de 10 Unidades E o preço maior que 20 

dados  = {
    'Medicamento':['Dipirona','Dorflex','Omeprazol','Neosaldina'],
    'Unidade':[4,3,15,19],
    'Preço' : [4.50,25,22,18.25]

}

df = pd.DataFrame(dados)
filtro = df[(df['Unidade'] < 10) & (df['Preço']>20)] 
filtro
```

```python
# filtrar Alunos cujo a Idade seja Maior que 20 e o Curso seja "TI"

dados = {
    'Nome' : ['João','Maria','Ana','Roberta','Letícia'],
    'Curso' : ['TI','Artes','Direito','Pedagogia','Ed.Física'],
    'Idade' : [25,28,35,42,18]
}

df = pd.DataFrame(dados)
filtro = df[(df['Idade']>20) & (df['Curso'] =='TI')] 
filtro
```
