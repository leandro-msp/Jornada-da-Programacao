Utilizando Pandas para extrair MÉDIA, MEDIANA E MODA 

```python
import pandas as pd
```

```python
#Criando Lista de Dados

dados = {
    'Marca' : ['Dell','HP','Lenovo','Apple'],
    'Modelos' : ['Inspiron 15','Pavilion','Thinkpad','Macbook Air'],
    'Armazenamento (GB)' : ['500 GB', '700 GB','1TB','2TB'],
    'RAM (GB)' :['120GB','250GB','500GB','700GB']
}

df = pd.DataFrame(dados)
df ['Quantidade'] = [15,30,50,40]
df
```

```python
dados = {
    'Produto' : ['Celular','Computador','Notebook','Videogame'],
    'Quantidade' :[6,5,20,4],
    'Preço' : [2500,4000,2200,5000],
    'Data' : ['23/03','21/03','20/03','01/01']
}

df = pd.DataFrame(dados)
df ['Total'] = (df['Quantidade'] * df['Preço']) # Cria uma nova coluna com nome "Total" , sendo o seu valor a multiplicação da Quantidade com o Preço
#df <-- print

df.rename(columns={'Quantidade':'Quantidade Vendida'}, inplace= True) # renomeia o nome da coluna | Inplace=True --> a alteração é feita diretamente do dataframe original
df
```

```python
dados= {
    'Nomes' : ['Ana','Fabrício','Juliana', 'Romero Brito'],
    'Altura' : [1.50,2.10,1.65,1.80]
}

df = pd.DataFrame(dados)

média = df['Altura'].mean()  # média --> soma todo os valores, e divide o resultado pela quantidade de elementos
mediana = df['Altura'].median() # median -->  pega o valor que está localizado no centro | se qtd de elementos for Par, ele soma os dois centrais e divide por 2.
moda = df['Altura'].mode() # apresenta o elemento que mais se repete, caso nn haja repetição ele pega o primeiro valor


print("A média é: ",média)
print("A mediana é :" , mediana)
print("a moda é : " , moda)
```

```python
dados = {
    'Nome':['Robson','Pietro','Diego','Rafael'],
    'Cargo':['Presidente','Diretor','Analista','Estagiário'],
    'Salário':[15000,7500,5000,1300]
}
df = pd.DataFrame(dados)

média = df['Salário'].mean()
print("Valor da média:" , média)

mediana = df ['Salário'].median()
print ("Valor da Mediana: ",mediana)

moda = df['Salário'].mode()
print("Valor da moda: " , moda)
```

```python
dados = {
    'Dia' :['Segunda','Terça','Quarta','Quinta','Sexta'],
    'Temperatura': [21,19,24,27,16]

}

df = pd.DataFrame(dados)

média = df ['Temperatura'].mean()
print("A temperatura média é: ", média)

moda = df ['Temperatura'].mode()
print("A moda entre as temperaturas é:" , moda)


mediana = df ['Temperatura'].median()
print("A mediana é:" , mediana)

```

```python
dados = {
    'Preços' : [200,300,500,200,600,150]
}

df = pd.DataFrame(dados)

media = df['Preços'].mean()
mediana =  df['Preços'].median()
moda =  df['Preços'].mode()

print(media)
print(mediana)
print(moda)
```
