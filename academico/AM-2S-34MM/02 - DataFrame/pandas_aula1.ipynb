```python
import pandas as pd   # importando a biblioteca pandas para o uso das funções.
```

```python
#Criação de uma lista

data = {'Nome': ['João','Maria','José'],
        'Idade': [20,18,35],
        'Cidade': ['Pinda','Paraisópolis','Itaquera']
}
# a variável 'data' conterá os valores da lista

df = pd.DataFrame(data) # chama a função DataFrame da biblioteca Pandas,para fazer a leitura dos dados da variável 'data' e atribuir para á variável 'df'
df
```

```python
# Adicionando nova coluna com novos dados

df ={'Nomes': ['Bernardo','Fátima','Henrique'],
     'Idade':[20,40,60]
     }

df = pd.DataFrame(df)
df ['Notas'] = [7,8,10] # adiciona uma nova coluna chamada "Notas" com os valores 7,8,10
df
```

```python
#APAGANDO COLUNA

#criando novo dataframe a partir do anterior, mas com a remação de uma coluna específicas

df_alunos_sem_idade = df.drop('Idade',axis=1)  # . drop --> deleta coluna | axis=1  -- > argumento que específica que ação é para fazer em colunas e vez de apagar linhas(dado único correspondente ao nome fornecido)
print("DataFrame sem a coluna 'Idade' :")
print (df_alunos_sem_idade) 
```
