```python
#Verificar se número é ímpar ou par, com base no resto da divisão.

numero = int(input("Digite um número: "))
if numero%2 == 0:
  print("o número escolhido é par")
else:
   print("o número escolhido é impar")
```

```python
#Verificar o status do Aluno conforme as Notas.

nota1= float(input("Digite a nota 1: "))
nota2= float(input("Digite a nota 2: "))
media= (nota1+nota2) / 2

if media >=7:
  print("APROVADO!!")
elif media >= 5 and media <=6.9: # elif é utilizado para criar vários caminhos de decisões, típico quando há mais de 2 condições a serem analisadas.
  print("Recuperação!")
else:
  print ("Reprovado")
print (f"Média: {media}") # o f-string permite que adicione variáveis diretamente dentro do texto.
```

```python
# Verificação de Faixa Etária

idade = int(input("Digite a idade: "))

if idade <12:
  print("vc Criança.")
elif idade >=12 and idade <=17:
  print("Vc é adolescente!")
elif idade >= 19 and idade <=59:
  print("Vc é adulto.")
else:
  print("Vc é idoso!")
```
